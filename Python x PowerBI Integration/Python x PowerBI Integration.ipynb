```python
# Importing all libraries needed
import pandas as pd
import seaborn as sns
```

```python
# Importing dataset
df = pd.read_csv('C:/Users/Dell/Documents/GitHub/Python-Learning-Journey/Python x PowerBI Integration/expenses.csv')

# Display the first five columns to have a better understanding of our dataset
print(df.head())

# Check the number of rows and columns
print("Dataset shape:", df.shape)

# Display the info about our dataset such as column names, how many values, dataset and data types
df.info()

# Display how many duplicated values in our dataset
print("Duplicated values = ", df.duplicated().sum())

# Now, we are able to see that this dataset contains info about my expenses 
```

```python
# Summarize it to make easier to have more data about the data

summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isna().sum(),
    "Unique Values": df.nunique()
})

display(summary)
```

```python
# Since some of the values are different, we will fix it now

# Here we will start fixing the date column making all values in a single format

df["date"] = pd.to_datetime(
    df["date"],
    format = "mixed",
    dayfirst = False,
    errors = "coerce"
)

# Print the "date" column 
print(df["date"])

print(df["date"].dtype)
```

```python
# Now we move to the next column

# Display key-values in "category"
print(df["category"].unique())
```

```python
# First, remove any extra spaces within the ''

df["category"] = df["category"].str.strip().str.lower()

# Here we will map our categories to make easier

map_category = {  
    "medical": "Health",
    "health": "Health",

    "food": "Groceries",
    "groceries": "Groceries",
    "grocery": "Groceries",

    "dining": "Eating Out",
    "eating out": "Eating Out",
    "restaurants": "Eating Out",

    "transport": "Transport",
    "uber/lyft": "Transport",

    "rent": "Housing",
    "housing": "Housing",

    "bills": "Utilities",
    "utilities": "Utilities",

    "shopping": "Shopping",

    "entertainment": "Entertainment",
    "fun": "Entertainment",

    "travel": "Travel",
    "amazon": "Shopping"
}

# Here we will apply the dictionary to replace the "category" values, using the 'replace' function
df["category"] = df["category"].replace(map_category)

# Print our new dataset with "category" updated
print(df.head())
```

```python
# Display key-values in "merchant"
print(df["merchant"].unique())

# We can see 'nan' so we have identified some empty values there
# To fix it, I will replace any wrong values to "Unknown"
df["merchant"] = df["merchant"].fillna("Unknown")
print(df["merchant"].unique())
```

```python
# Above we compare the fixed "merchant" column, now we move to "amount"

# Display key-values in "amount"

print(df["amount"].head(10))
print("Data type:", df["amount"].dtype)
print("Missing:", df["amount"].isna().sum())
print(df["amount"].describe())
```

```python
# Set all values in "amount" column to a single one
df["amount"] = pd.to_numeric(
    df["amount"], 
    errors="coerce")

# Now we check for negative values, those are refunds and I will use them on the last step
print(df[df["amount"] < 0])

# Payment method is all good so no need to fix there
```

```python
# For notes, we have a couple of missing values and lower case

# First we replace any missing values to "None"
df["notes"] = df["notes"].fillna("None")

# Fixing the lower case issue
df["notes"] = df["notes"].str.strip().str.title()

print(df["notes"].head())
```

```python
# Remove rows missing essential information
df = df.dropna(subset=["date", "amount", "category"])

# Display the info about our dataset such as column names, how many values, dataset and data types
print(df.info())
```

```python
# Before saving it as a CSV and moving to PowerBI, lets fix the payments and the refunds

# Purchases
purchases = df[df["amount"] > 0]

# Refunds
refunds = df[df["amount"] < 0]

# Calculate totals
total_spending = purchases["amount"].sum()
total_refunds = abs(refunds["amount"].sum()) #abs function used to display refund as positive value but remains negative in dataset
net_spending = df["amount"].sum()

print(f"Total purchases: €{total_spending:.2f}")
print(f"Total refunds: €{total_refunds:.2f}")
print(f"Net spending: €{net_spending:.2f}")
```

```python
 # Save the cleaned dataset
df.to_csv('expenses_dataset_clean.csv', index=False)
```
